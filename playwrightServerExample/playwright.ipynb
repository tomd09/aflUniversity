```python
import requests
```

```python
#### using a server to make async playwright requests, can then be scraped with bs4

apiURL = 'http://127.0.0.1:8000/scrape'

def fetchViaApi(url):
    response = requests.get(apiURL, params={'url': url})
    response.raise_for_status()
    data = response.json()
    return data['html']

url = 'https://www.playhq.com/afl/org/university-perth-football-league/97522c61/perth-football-league-2025/teams/university-c3/bb882c71'
html = fetchViaApi(url)
print(html)
```
